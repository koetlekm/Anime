```python
# === 1.4: Data Cleaning & Merging ===

"""
Rationale: This helper function handles potential duplicates in the static 'trees' data.
If duplicates have different biochemical data, data is averaged - else, take the first entry only.
This ensures one unique record per tree_id and prints its actions.
"""


def resolve_duplicates_simple(
    trees_df: pd.DataFrame, key_col="tree_id", name_col="tree_name", agg="mean"
):
    trees_df = trees_df.copy()
    total_rows_before = len(trees_df)
    dup_mask = trees_df[key_col].duplicated(keep=False)

    print(f"\n--- Duplicate Resolution ({key_col}) ---")
    if not dup_mask.any():
        print(f"Total rows: {total_rows_before}. No duplicates found. OK.")
        return trees_df

    total_duplicate_rows = dup_mask.sum()
    duplicate_ids = trees_df.loc[dup_mask, key_col].unique()
    total_duplicate_ids = len(duplicate_ids)

    bio_cols = [c for c in trees_df.columns if c not in [key_col, name_col]]
    num_bio_cols = [c for c in bio_cols if pd.api.types.is_numeric_dtype(trees_df[c])]
    dup_df = trees_df.loc[dup_mask].sort_values([key_col])

    varying_ids = []
    if num_bio_cols:
        varying = (
            dup_df.groupby(key_col)[num_bio_cols]
            .nunique(dropna=False)
            .max(axis=1)
            .gt(1)
        )
        varying_ids = varying[varying].index.tolist()

    n_averaged_ids = len(varying_ids)
    n_simple_dropped_ids = total_duplicate_ids - n_averaged_ids

    print(f"Total rows before: {total_rows_before}")
    print(
        f"Found {total_duplicate_rows} rows belonging to {total_duplicate_ids} duplicate IDs."
    )
    print(
        f"  > {n_averaged_ids} IDs had conflicting numeric data and will be **averaged**."
    )
    print(
        f"  > {n_simple_dropped_ids} IDs had identical data; **keeping first entry**."
    )

    if n_averaged_ids == 0:
        trees_clean = trees_df.drop_duplicates(
            subset=[key_col], keep="first"
        ).reset_index(drop=True)
    else:
        agg_num = {c: agg for c in num_bio_cols}
        non_num = [c for c in bio_cols if c not in num_bio_cols]
        agg_non = {c: "first" for c in non_num}
        trees_clean = (
            trees_df.groupby([key_col, name_col], as_index=False)
            .agg({**agg_num, **agg_non})
            .reset_index(drop=True)
        )
        cols = [key_col, name_col] + [c for c in bio_cols if c in trees_clean.columns]
        trees_clean = trees_clean[cols]


    total_rows_after = len(trees_clean)
    print(
        f"Total rows after: {total_rows_after} (Resolved {total_rows_before - total_rows_after} rows)"
    )
    return trees_clean


# --- 1. Clean TREES.pkl ---
trees = resolve_duplicates_simple(trees_raw)
bio_cols = [c for c in trees.columns if c not in ["tree_id", "tree_name"]]
trees[bio_cols] = trees[bio_cols].apply(pd.to_numeric, errors="coerce")
trees["tree_name"] = trees["tree_name"].astype("category")

# --- 2. Clean TREE_EVENTS.pkl ---
ev = tree_events_raw.copy()
ev["tree_id"] = pd.to_numeric(ev["tree_id"], errors="coerce", downcast="integer")
ev["day"] = pd.to_numeric(ev["day"], errors="coerce", downcast="integer")
ev["node_failure"] = pd.to_numeric(
    ev["node_failure"], errors="coerce", downcast="integer"
)
for c in ["humidity", "pollution_index", "tree_health_index"]:
    ev[c] = pd.to_numeric(ev[c], errors="coerce")

# Rationale: Drop rows where key identifiers are missing.
ev.dropna(subset=["tree_id", "day"], inplace=True)
# Rationale: Drop duplicate (tree_id, day) entries.
ev.drop_duplicates(subset=["tree_id", "day"], keep="first", inplace=True)

# --- 3. Merge Datasets ---
# Rationale: Inner join. We only want events for trees we have biometric data for.
valid_tree_ids = set(trees["tree_id"])
ev = ev[ev["tree_id"].isin(valid_tree_ids)]

df_master = ev.merge(trees, on="tree_id", how="left", validate="m:1")
df_master.sort_values(["tree_id", "day"], inplace=True)
df_master.reset_index(drop=True, inplace=True)

print(f"\nFinal merged data shape: {df_master.shape}")
print("--- Task 1 (Data Prep) Complete ---")
```

```python
# Step 3 — Column-level data-quality function
def col_quality(series: pd.Series) -> dict:
    n = len(series)
    miss = series.isna().sum()
    miss_frac = miss / n

    is_num = pd.api.types.is_numeric_dtype(series)
    stats = {
        'count_non_missing': int(n - miss),
        'missing_count': int(miss),
        'missing_frac': float(miss_frac),
        'dtype': str(series.dtype),
        'n_unique_non_missing': int(series.nunique(dropna=True)),
    }

    if is_num:
        s = series.dropna()
        stats.update({
            'min': float(s.min()) if len(s) else np.nan,
            'p25': float(s.quantile(0.25)) if len(s) else np.nan,
            'median': float(s.median()) if len(s) else np.nan,
            'mean': float(s.mean()) if len(s) else np.nan,
            'p75': float(s.quantile(0.75)) if len(s) else np.nan,
            'max': float(s.max()) if len(s) else np.nan,
            'std': float(s.std(ddof=1)) if len(s) > 1 else np.nan,
            'skew': float(s.skew()) if len(s) > 1 else np.nan,
            'kurtosis': float(s.kurtosis()) if len(s) > 1 else np.nan,
        })
        # Domain checks for biomarker masses (µg): should be >= 0
        stats['negative_count'] = int((s < 0).sum())
        stats['zero_count'] = int((s == 0).sum())
        stats['zero_frac'] = float((s == 0).mean()) if len(s) else np.nan

        # IQR outlier detection (Tukey rule)
        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1
        if pd.notna(iqr) and iqr > 0:
            lower = q1 - 1.5 * iqr
            upper = q3 + 1.5 * iqr
            out_ct = int(((s < lower) | (s > upper)).sum())
            stats['iqr_outlier_count'] = out_ct
            stats['iqr_outlier_frac'] = float(out_ct / len(s))
        else:
            stats['iqr_outlier_count'] = np.nan
            stats['iqr_outlier_frac'] = np.nan

        # Near-zero variance flag (simple heuristic)
        std = stats['std']
        nzv = (stats['n_unique_non_missing'] <= 1) or (isinstance(std, float) and std == 0.0)
        stats['near_zero_variance'] = bool(nzv)
    else:
        stats['negative_count'] = np.nan
        stats['zero_count'] = np.nan
        stats['zero_frac'] = np.nan
        stats['iqr_outlier_count'] = np.nan
        stats['iqr_outlier_frac'] = np.nan
        stats['near_zero_variance'] = False

    return stats
```

```python
# === 2.1: Regression Feature Engineering ===
print("\n=== STARTING TASK 2: REGRESSION (Tree Health) ===")

df_reg = df_master.copy()
g_reg = df_reg.groupby("tree_id", group_keys=False)
sensor_cols = ["humidity", "pollution_index"]
for w in [5, 10]:
    for c in sensor_cols:
        df_reg[f"{c}_roll_avg_{w}"] = (
            g_reg[c].rolling(w, min_periods=1).mean().reset_index(level=0, drop=True)
        )
        df_reg[f"{c}_roll_std_{w}"] = (
            g_reg[c].rolling(w, min_periods=1).std().reset_index(level=0, drop=True)
        )

# --- CRITICAL FEATURE: Add 1-day lag of the target variable ---
print("Adding health_index_lag1 feature...")
df_reg["health_index_lag1"] = g_reg["tree_health_index"].shift(1)
```

```python
# ===== Triage Base Models; Linear, Spline, Polynomial and Trees =====
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.compose import make_column_selector as selector
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.model_selection import GroupKFold, KFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (OneHotEncoder, PolynomialFeatures,
                                   QuantileTransformer, SplineTransformer,
                                   StandardScaler)


def _onehot_dense():
    try:
        return OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", sparse=False)


def _cv(model, X, y, cv, groups=None):
    s = cross_val_score(model, X, y, cv=cv, groups=groups, scoring="r2", n_jobs=-1)
    return float(np.mean(s)), float(np.std(s))


def triage_v3(X, y, groups=None, n_splits=5, random_state=42):
    num_sel = selector(dtype_include=np.number)
    cat_sel = selector(dtype_exclude=np.number)

    # Linear-friendly preprocessing (scale numerics, OHE categoricals)
    pre_lin = ColumnTransformer(
        [
            (
                "num",
                Pipeline(
                    [
                        ("imputer", SimpleImputer(strategy="median")),
                        ("scaler", StandardScaler()),
                    ]
                ),
                num_sel,
            ),
            (
                "cat",
                Pipeline(
                    [
                        ("imputer", SimpleImputer(strategy="most_frequent")),
                        (_onehot_dense().__class__.__name__.lower(), _onehot_dense()),
                    ]
                ),
                cat_sel,
            ),
        ]
    )

    # Spline setup
    pre_gam = ColumnTransformer(
        [
            (
                "num",
                Pipeline(
                    [
                        ("imputer", SimpleImputer(strategy="median")),
                        (
                            "q",
                            QuantileTransformer(
                                output_distribution="uniform",
                                subsample=200000,
                                random_state=random_state,
                            ),
                        ),
                        # --- FIX: Changed extrapolation to 'constant' to prevent explosion ---
                        (
                            "spline",
                            SplineTransformer(
                                n_knots=5, degree=3, extrapolation="constant"
                            ),
                        ),
                        ("scaler", StandardScaler()),
                    ]
                ),
                num_sel,
            ),
            (
                "cat",
                Pipeline(
                    [
                        ("imputer", SimpleImputer(strategy="most_frequent")),
                        (_onehot_dense().__class__.__name__.lower(), _onehot_dense()),
                    ]
                ),
                cat_sel,
            ),
        ]
    )

    # Tree-friendly preprocessing (impute, OHE)
    pre_tree = ColumnTransformer(
        [
            ("num", Pipeline([("imputer", SimpleImputer(strategy="median"))]), num_sel),
            (
                "cat",
                Pipeline(
                    [
                        ("imputer", SimpleImputer(strategy="most_frequent")),
                        (_onehot_dense().__class__.__name__.lower(), _onehot_dense()),
                    ]
                ),
                cat_sel,
            ),
        ]
    )

    # === Define Model Families ===

    # 1. Simple Linear
    ridge = TransformedTargetRegressor(
        regressor=Pipeline(
            [("pre", pre_lin), ("ridge", Ridge(random_state=random_state))]
        ),
        transformer=StandardScaler(),
    )

    # 2. Additive Non-Linear (Splines)
    gam_ridge = TransformedTargetRegressor(
        regressor=Pipeline(
            [("pre", pre_gam), ("ridge", Ridge(random_state=random_state))]
        ),
        transformer=StandardScaler(),
    )

    # 3. *** NEW: Interaction Non-Linear (Polynomial) ***
    poly_ridge = TransformedTargetRegressor(
        regressor=Pipeline(
            [
                ("pre", pre_lin),
                ("poly", PolynomialFeatures(degree=2, include_bias=False)),
                ("ridge", Ridge(random_state=random_state)),
            ]
        ),
        transformer=StandardScaler(),
    )

    # 4. Tree-based (Interactions)
    hgb = TransformedTargetRegressor(
        regressor=Pipeline(
            [
                ("pre", pre_tree),
                # --- FIX: Use robust defaults for triage ---
                ("hgb", HistGradientBoostingRegressor(random_state=random_state)),
            ]
        ),
        transformer=StandardScaler(),
    )

    cv = (
        GroupKFold(n_splits=n_splits)
        if groups is not None
        else KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    )

    scores = {}
    models_to_run = {
        "Linear_Ridge": ridge,
        "Spline_Ridge": gam_ridge,
        "Polynomial_Ridge": poly_ridge,  # --- Added new model ---
        "HGB_Tree": hgb,
    }

    for name, m in models_to_run.items():
        try:
            mean, sd = _cv(m, X, y, cv, groups)
            scores[name] = {"mean": mean, "std": sd}
        except Exception as e:
            print(f"Model {name} failed: {e}")
            scores[name] = {"mean": -np.inf, "std": np.inf}

    # === Updated Recommendation Logic ===
    r_lin = scores.get("Linear_Ridge", {}).get("mean", -np.inf)
    r_gam = scores.get("Spline_Ridge", {}).get("mean", -np.inf)
    r_poly = scores.get("Polynomial_Ridge", {}).get("mean", -np.inf)
    r_hgb = scores.get("HGB_Tree", {}).get("mean", -np.inf)

    best = max(r_lin, r_gam, r_poly, r_hgb)

    shortlist, rationale = [], []

    # Check if Poly or HGB (interaction models) are the best
    if r_poly >= best - 0.03:
        shortlist = ["Polynomial(2-3)+ElasticNet/LinearSVR"]
        rationale.append(
            "Polynomial features (interactions) provide the strongest lift."
        )
    elif r_hgb >= best - 0.03:
        shortlist = ["GradientBoostedTrees (HGB/LightGBM/XGBoost)"]
        rationale.append("Tree-based interactions & thresholds drive lift.")
    # Check if Splines are best
    elif r_gam >= best - 0.03:
        shortlist = ["Spline+Ridge (GAM-like)"]
        rationale.append(
            "Smooth non-linear curvature helps; interactions not required."
        )
    # Fallback to linear
    else:
        shortlist = ["ElasticNet/Ridge", "LinearSVR"]
        rationale.append(
            "Linear baseline at/near the top; complex models not justified."
        )

    return {
        "cv_scores": scores,
        "recommendation": {"family": shortlist, "why": "; ".join(rationale)},
    }


# ---- Run
# (Assuming X_reg, y_reg, and groups_reg are already defined)
report_v3 = triage_v3(X_reg, y_reg, groups=groups_reg, n_splits=5)

print("\n--- Triage v3 Results ---")
print("CV R2:")
for k, v in report_v3["cv_scores"].items():
    print(f"  {k:<18}: {v['mean']:.4f} (±{v['std']:.4f})")

print("\nShortlist:", report_v3["recommendation"]["family"])
print("Why:", report_v3["recommendation"]["why"])

from sklearn.compose import make_column_selector as selector
```

```python
# === 3. Task 3: Understanding Device Failure (Classification) ===
# Rationale: This task requires us to understand and predict device failure
# to help plan service actions. This is a classification problem.
# We will use the 7-day-ahead classification script provided.

print("\n=== STARTING TASK 3: CLASSIFICATION (Device Failure) ===")

# -------------------------
# 3.1) Feature Engineering
# -------------------------
# Rationale: We engineer features based on past data to predict the future.
# This includes lags (what happened yesterday), rolling windows (recent trends),
# and time-since-event (days since last failure).
df_cls = df_master.copy()
g_cls = df_cls.groupby("tree_id", group_keys=False)

# Rationale: Target definition. We are predicting if a failure will
# occur ANY time in the next 7 days (t+1 to t+7). This is a
# practical, actionable definition for maintenance planning.
df_cls["node_failure_7d"] = (
    g_cls["node_failure"].shift(-1).rolling(7, min_periods=1).max()
)

# Lags
df_cls["node_failure_lag1"] = g_cls["node_failure"].shift(1)
df_cls["humidity_lag1"] = g_cls["humidity"].shift(1)
df_cls["pollution_lag1"] = g_cls["pollution_index"].shift(1)
df_cls["health_lag1"] = g_cls["tree_health_index"].shift(1)

# Rolling features (7-day window, must be full)
for col in ["humidity", "pollution_index", "tree_health_index"]:
    df_cls[f"{col}_rollmean_7"] = (
        g_cls[col].rolling(7, min_periods=7).mean().reset_index(level=0, drop=True)
    )
    df_cls[f"{col}_rollstd_7"] = (
        g_cls[col].rolling(7, min_periods=7).std().reset_index(level=0, drop=True)
    )
    df_cls[f"{col}_delta_7"] = df_cls[col] - df_cls[f"{col}_rollmean_7"]


# Days since last failure
def days_since_last_fail(group):
    last = -np.inf
    out = []
    for d, f in zip(group["day"].values, group["node_failure"].values):
        out.append(d - last if np.isfinite(last) else np.nan)
        if f == 1:
            last = d
    return pd.Series(out, index=group.index)


df_cls["days_since_fail"] = g_cls.apply(days_since_last_fail)

# -------------------------
# 3.2) Final Data Assembly
# -------------------------
TARGET_CLS = "node_failure_7d"

# Rationale: Drop all rows with NaNs generated by the rolling/lag features.
# This ensures models only train on complete feature sets.
essential_features = [
    TARGET_CLS,
    "node_failure_lag1",
    "humidity_lag1",
    "pollution_lag1",
    "health_lag1",
    "humidity_rollmean_7",
    "pollution_index_rollmean_7",
    "tree_health_index_rollmean_7",
    "days_since_fail",
]
have = [c for c in essential_features if c in df_cls.columns]
df_cls.dropna(subset=have, inplace=True)

y_cls = df_cls[TARGET_CLS].astype(int)
groups_cls = df_cls["tree_id"]

# Rationale: Drop the target and identifiers, AND the raw 'node_failure'
# column to prevent leaking the current day's failure status.
drop_cols_cls = {TARGET_CLS, "node_failure", "tree_id", "day"}
X_cls = df_cls.drop(columns=[c for c in drop_cols_cls if c in df_cls.columns]).copy()

if "tree_name" in X_cls.columns:
    X_cls["tree_name"] = X_cls["tree_name"].astype("string")

print(f"Classification X: {X_cls.shape} | y: {y_cls.shape}")
cls_counts = np.bincount(y_cls)
print("Class balance:", cls_counts, "=>", (cls_counts / y_cls.size).round(3))

# Identify feature types
cat_features_cls = X_cls.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()
num_features_cls = X_cls.select_dtypes(include=[np.number]).columns.tolist()


# -------------------------
# 3.3) Custom StratifiedGroupKFold
# -------------------------
# Rationale: This is a custom CV strategy. We must keep groups (trees)
# separate (like GroupKFold), but the 'node_failure_7d' target is
# highly imbalanced. This function tries to balance the number
# of positive samples across all 5 folds, giving a more stable
# and reliable metric.
def stratified_group_kfold(y, groups, n_splits=5, random_state=42):
    df_aux = (
        pd.DataFrame({"group": groups, "y": y})
        .groupby("group")["y"]
        .agg(["size", "sum"])
        .reset_index()
    )
    df_aux = df_aux.sample(frac=1.0, random_state=random_state).reset_index(drop=True)
    fold_pos = np.zeros(n_splits, dtype=int)
    fold_sizes = np.zeros(n_splits, dtype=int)
    group_to_fold = {}
    df_aux = df_aux.sort_values("sum", ascending=False).reset_index(drop=True)
    for _, row in df_aux.iterrows():
        g_id, g_n, g_pos = row["group"], int(row["size"]), int(row["sum"])
        scores = [(fold_pos[k], fold_sizes[k], k) for k in range(n_splits)]
        scores.sort()
        best_k = scores[0][2]
        group_to_fold[g_id] = best_k
        fold_pos[best_k] += g_pos
        fold_sizes[best_k] += g_n
    idx = np.arange(len(y))
    for k in range(n_splits):
        val_groups = [g for g, fk in group_to_fold.items() if fk == k]
        val_mask = groups.isin(val_groups).values
        train_idx = idx[~val_mask]
        val_idx = idx[val_mask]
        yield (train_idx, val_idx)


# Precompute our group-stratified splits
splits_cls = list(
    stratified_group_kfold(y=y_cls, groups=groups_cls, n_splits=5, random_state=42)
)

# -------------------------
# 3.4) Preprocessing Pipelines
# -------------------------
# Rationale: We define 3 distinct preprocessors inside pipelines to
# prevent data leakage from the validation set.
# 1. Linear/SVM: Impute + Scale (Numeric), Impute + OHE (Categorical)
# 2. Trees (RF/XGB): Impute (Numeric), Impute + OHE (Categorical)
# 3. CatBoost: Impute (Numeric), Impute (Categorical) - CatBoost handles cats natively.

preprocessor_linear_cls = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline(
                steps=[
                    ("imp", SimpleImputer(strategy="median")),
                    ("scaler", StandardScaler()),
                ]
            ),
            num_features_cls,
        ),
        (
            "cat",
            Pipeline(
                steps=[
                    ("imp", SimpleImputer(strategy="constant", fill_value="missing")),
                    (
                        "ohe",
                        OneHotEncoder(
                            handle_unknown="ignore", drop="first", sparse_output=False
                        ),
                    ),
                ]
            ),
            cat_features_cls,
        ),
    ],
    remainder="drop",
)

preprocessor_tree_cls = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline(steps=[("imp", SimpleImputer(strategy="median"))]),
            num_features_cls,
        ),
        (
            "cat",
            Pipeline(
                steps=[
                    ("imp", SimpleImputer(strategy="constant", fill_value="missing")),
                    (
                        "ohe",
                        OneHotEncoder(
                            handle_unknown="ignore", drop="first", sparse_output=False
                        ),
                    ),
                ]
            ),
            cat_features_cls,
        ),
    ],
    remainder="drop",
)

preprocessor_cb_cls = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline(steps=[("imp", SimpleImputer(strategy="median"))]),
            num_features_cls,
        ),
        (
            "cat",
            Pipeline(
                steps=[
                    ("imp", SimpleImputer(strategy="constant", fill_value="missing"))
                ]
            ),
            cat_features_cls,
        ),
    ],
    remainder="drop",
)


# -------------------------
# 3.5) Evaluation Helpers
# -------------------------
def evaluate_folds(y_true_list, proba_list, pred_list):
    """Aggregate standard metrics over folds."""
    ROC, PR, BACC, F1, BRIER = [], [], [], [], []
    for yt, pb, pd_ in zip(y_true_list, proba_list, pred_list):
        ROC.append(roc_auc_score(yt, pb) if len(np.unique(yt)) > 1 else np.nan)
        PR.append(average_precision_score(yt, pb))
        BACC.append(balanced_accuracy_score(yt, pd_))
        F1.append(f1_score(yt, pd_, zero_division=0))
        BRIER.append(brier_score_loss(yt, pb))

    def stat(arr):
        return (np.nanmean(arr), np.nanstd(arr))

    return {
        "roc_auc(mean±sd)": stat(ROC),
        "pr_auc(mean±sd)": stat(PR),
        "balanced_acc(mean±sd)": stat(BACC),
        "f1(mean±sd)": stat(F1),
        "brier(mean±sd)": stat(BRIER),
    }


def print_report(name, report):
    def fmt(tup):
        return f"{tup[0]:.4f} (±{tup[1]:.4f})"

    print(f"\n--- {name} ---")
    # Rationale: We focus on PR_AUC (good for imbalance) and ROC_AUC (general rank skill).
    for k, v in report.items():
        print(f"{k:<24} : {fmt(v)}")


# Master evaluation function
def run_pipeline_cv(name, build_estimator_fn):
    yt_list, pb_list, pd_list = [], [], []
    for fold, (tr, va) in enumerate(splits_cls, 1):
        X_tr, X_va = X_cls.iloc[tr], X_cls.iloc[va]
        y_tr, y_va = y_cls.iloc[tr], y_cls.iloc[va]

        # Rationale: build_estimator_fn allows us to calculate class
        # weights *only* on the fold's training data (y_tr),
        # which is critical for handling imbalance correctly.
        model = build_estimator_fn(y_tr)
        model.fit(X_tr, y_tr)

        proba_va = model.predict_proba(X_va)[:, 1]
        pred_va = (proba_va >= 0.5).astype(int)

        yt_list.append(y_va.values)
        pb_list.append(proba_va)
        pd_list.append(pred_va)

    report = evaluate_folds(yt_list, pb_list, pd_list)
    print_report(name, report)


# -------------------------
# 3.6) Run Classification Models
# -------------------------

# ---- Baseline: Dummy (most frequent) ----
run_pipeline_cv(
    "Dummy (most_frequent)", lambda yy: DummyClassifier(strategy="most_frequent")
)


# ---- Logistic Regression ----
def build_logreg(y_tr):
    return Pipeline(
        [
            ("pre", preprocessor_linear_cls),
            (
                "clf",
                LogisticRegression(
                    solver="liblinear",
                    class_weight="balanced",
                    max_iter=1000,
                    random_state=42,
                ),
            ),
        ]
    )


run_pipeline_cv("Logistic Regression", build_logreg)


# ---- Random Forest ----
def build_rf(y_tr):
    return Pipeline(
        [
            ("pre", preprocessor_tree_cls),
            (
                "clf",
                RandomForestClassifier(
                    n_estimators=400,
                    max_depth=None,
                    min_samples_leaf=5,
                    class_weight="balanced_subsample",
                    n_jobs=-1,
                    random_state=42,
                ),
            ),
        ]
    )


run_pipeline_cv("Random Forest", build_rf)


# ---- XGBoost (per-fold scale_pos_weight) ----
def build_xgb(y_tr):
    bc = np.bincount(y_tr)
    spw = (bc[0] / bc[1]) if len(bc) > 1 and bc[1] > 0 else 1.0
    return Pipeline(
        [
            ("pre", preprocessor_tree_cls),
            (
                "clf",
                XGBClassifier(
                    n_estimators=500,
                    learning_rate=0.05,
                    max_depth=5,
                    subsample=0.9,
                    colsample_bytree=0.9,
                    objective="binary:logistic",
                    eval_metric="aucpr",
                    scale_pos_weight=spw,  # Rationale: Key for imbalance
                    n_jobs=-1,
                    random_state=42,
                ),
            ),
        ]
    )


run_pipeline_cv("XGBoost", build_xgb)


# ---- CatBoost (native categoricals) ----
# Rationale: CatBoost needs a special wrapper to get the
# categorical feature indices *after* the preprocessor (imputer) runs.
class CatBoostWithPre:
    def __init__(self):
        self.pre = preprocessor_cb_cls
        self.cb = None
        self.cols_ = None
        self.cat_idx_ = None

    def fit(self, Xf, yf):
        self.pre.fit(Xf, yf)
        Xp = self.pre.transform(Xf)
        self.cols_ = list(num_features_cls) + list(cat_features_cls)
        Xpd = pd.DataFrame(Xp, columns=self.cols_)
        self.cat_idx_ = [
            Xpd.columns.get_loc(c) for c in cat_features_cls if c in Xpd.columns
        ]
        self.cb = CatBoostClassifier(
            iterations=1000,
            learning_rate=0.05,
            depth=6,
            loss_function="Logloss",
            eval_metric="AUC",
            auto_class_weights="Balanced",  # Rationale: Key for imbalance
            random_seed=42,
            verbose=0,
            early_stopping_rounds=50,
        )
        self.cb.fit(Xpd, yf, cat_features=self.cat_idx_)
        return self

    def predict_proba(self, Xf):
        Xp = self.pre.transform(Xf)
        Xpd = pd.DataFrame(Xp, columns=self.cols_)
        return self.cb.predict_proba(Xpd)


run_pipeline_cv("CatBoost", lambda yy: CatBoostWithPre())

print("\n=== END: Base Classification Pipeline ===")
```

```python
# === 3.6: Tuning the Logistic Regression Model ===
from sklearn.model_selection import GridSearchCV, GroupKFold
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import make_column_selector as selector
import numpy as np

print("--- Starting GridSearchCV for Logistic Regression ---")

# --- 1. Re-define necessary helpers & selectors ---
def _onehot_dense():
    try: return OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    except TypeError: return OneHotEncoder(handle_unknown="ignore", sparse=False)

num_sel = selector(dtype_include=np.number)
cat_sel = selector(dtype_exclude=np.number)

# --- 2. Define the CV splitter ---
# We use GroupKFold to respect the 'tree_id' groups
gkf_cls = GroupKFold(n_splits=5)

# --- 3. Define the preprocessing pipeline ---
# This is the same preprocessor used for the linear model in the previous step
preprocessor_linear_cls = ColumnTransformer(
    transformers=[
        ('num', Pipeline(steps=[('imp', SimpleImputer(strategy='median')),
                                ('scaler', StandardScaler())]), num_sel),
        ('cat', Pipeline(steps=[('imp', SimpleImputer(strategy='constant', fill_value='missing')),
                                ('ohe', _onehot_dense())]), cat_sel),
    ], remainder='drop'
)

# --- 4. Create the full model pipeline ---
lr_pipeline = Pipeline([
    ('pre', preprocessor_linear_cls),
    ('clf', LogisticRegression(solver='liblinear', # liblinear works well with both l1/l2
                                max_iter=1000,
                                random_state=RANDOM_STATE))
])

# --- 5. Define the Hyperparameter Grid to Search ---
param_grid_lr = {
    # 'clf__penalty': Test both L1 (Lasso) and L2 (Ridge) regularization
    'clf__penalty': ['l1', 'l2'],

    # 'clf__C': Test a range of regularization strengths (log-scale)
    # Lower C = Stronger regularization
    'clf__C': np.logspace(-3, 2, 6),  # [0.001, 0.01, 0.1, 1, 10, 100]

    # 'clf__class_weight': Test if explicit balancing helps
    'clf__class_weight': ['balanced', None]
}

# --- 6. Set up and run the GridSearchCV ---
search_lr = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=param_grid_lr,
    scoring='balanced_accuracy',  # We optimize for the metric that mattered most
    cv=gkf_cls,
    n_jobs=-1,
    verbose=1
)

# Rationale: We must pass 'groups=groups_cls' to .fit()
# so the 'gkf_cls' splitter correctly uses the tree_id groups.
# (This assumes X_cls, y_cls, and groups_cls are in memory)
search_lr.fit(X_cls, y_cls, groups=groups_cls)

# --- 7. Print the results ---
print("\n--- Logistic Regression GridSearchCV Results ---")
print(f"Best Balanced Accuracy Score: {search_lr.best_score_:.4f}")
print(f"Best Parameters: {search_lr.best_params_}")
```
